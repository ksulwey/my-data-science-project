```python
# Импортируем библиотеки
import numpy as np
import pandas as pd

# Создаем случайную матрицу 100×5 (числа от 0 до 1)
data = np.random.rand(100, 5)

# Преобразуем в DataFrame для удобства
df = pd.DataFrame(data, columns=['Признак_1', 'Признак_2', 'Признак_3', 'Признак_4', 'Признак_5'])

# Выводим первые 5 строк, чтобы посмотреть на данные
print("Первые 5 строк матрицы:")
print(df.head())

# Выводим статистики (среднее, std, min, max, квартили)
print("\nСтатистики по каждому признаку:")
print(df.describe())
```
